```python
import random
def pedraPaperTisora():
  eines = ('pedra', 'paper', 'tisora')
  wJugador = 0
  wComp = 0

  while wJugador < 3 and wComp < 3:

    einaJugador = input('Introdueix que vols treure (pedra, paper, tisora): ').lower()

    einaComp = random.choice(eines)

    if einaJugador == einaComp:
      print('Empat!')
      print(f'Heu tret els dos {einaJugador}')

    elif (
        (einaJugador == 'pedra' and einaComp == 'tisora') or
        (einaJugador == 'tisora' and einaComp == 'paper') or
        (einaJugador == 'paper' and einaComp == 'pedra')):

      wJugador += 1
      print('Has guanyat!')
      print(f'Tu has tret {einaJugador} i la maquina {einaComp}')

    elif einaJugador not in eines:
      print('La opcio introduïda no es valida.')

    else:
      wComp += 1
      print('Has perdut!')
      print(f'Tu has tret {einaJugador} i la maquina {einaComp}')

    print(f'Contador victories (Millor de 3): \nJugador: {wJugador}, Maquina: {wComp}\n')
```

```python
def elPenjat():
  paraules = ['nube', 'gato', 'flor', 'reloj', 'piedra', 'bosque', 'luz', 'tren', 'playa', 'cielo', 'raton', 'fuego', 'libro', 'silla', 'dragon', 'verde', 'cobre', 'isla', 'fruta', 'sombras']

  paraulaSecreta = random.choice(paraules)
  contador = 0
  paraulaOculta = []

  for p in paraulaSecreta:
    paraulaOculta.append("_")

  while contador < (len(paraulaSecreta) * 2):

    print(" ".join(paraulaOculta))
    usuariLletra = input('\nIntrodueix una lletra: ').lower()

    if len(usuariLletra) > 1:
      print("Introdueix només una lletra")
    else:
      if usuariLletra in paraulaSecreta:
        for i in range(len(paraulaSecreta)):
          if paraulaSecreta[i] == usuariLletra:
            paraulaOculta[i] = usuariLletra

        print(f'Lletra {usuariLletra} correcta')
      else:
        contador += 1
        print(f'Lletra {usuariLletra} incorrecta')
        print(f'Et queden {len(paraulaSecreta) * 2 - contador} intents')

      if '_' not in paraulaOculta:
        print(f'\nHas guanyat!\nLa paraula era: {paraulaSecreta}')
        print(f"T'has equivocat {contador} vegades!")
        break

  if '_' in paraulaOculta:
      print(f"\nHas perdut! Màxim d'intents realitzat!")
      print(f'La paraula era: {paraulaSecreta}')
```

```python
tablero = ['1', '2', '3', '4', '5', '6', '7', '8', '9']

def mostrarTablero():
  fila = ""
  for i in range(len(tablero)):
    fila += tablero[i]

    if (i + 1) % 3 == 0:
      print(fila)
      fila = ""

      if i < 8:
        print("--+---+--")
    else:
      fila += " | "

def tresEnRatlla():
  jugador1 = "X"
  jugador2 = 'O'
  jugador = jugador1
  guanyador = False
  contador = 0

  mostrarTablero()

  while guanyador == False:
    posicio = input(f"Jugador {jugador}, tria una casella (1-9): ")

    if posicio in tablero:
      posicio = int(posicio) - 1

      if tablero[posicio] != "X" and tablero[posicio] != "O":
        tablero[posicio] = jugador

        if (tablero[0] == jugador and tablero[1] == jugador and tablero [2] == jugador) or \
           (tablero[0] == jugador and tablero[4] == jugador and tablero[8] == jugador) or \
           (tablero[6] == jugador and tablero[7] == jugador and tablero[8] == jugador) or \
           (tablero[3] == jugador and tablero[4] == jugador and tablero[5] == jugador) or \
           (tablero[2] == jugador and tablero[4] == jugador and tablero[6] == jugador) or \
           (tablero[1] == jugador and tablero[4] == jugador and tablero[7] == jugador) or \
           (tablero[2] == jugador and tablero[5] == jugador and tablero[8] == jugador) or \
           (tablero[0] == jugador and tablero[3] == jugador and tablero[6] == jugador):
           print(f'El jugador {jugador} ha guanyat!!')
           guanyador = True
           break;

      if jugador == jugador1:
        jugador = jugador2

      else:
        jugador = jugador1
      mostrarTablero()

    else:
      print('Posició incorrecta')
```

```python
def blackJack():
    sortir = False

    cartesCrupier = random.randint(1, 10) + random.randint(1, 10)
    cartesUsuari = random.randint(1, 10) + random.randint(1, 10)

    cartaVisible = random.randint(1, 10)
    cartaTapada = random.randint(1, 10)
    cartesCrupier = cartaVisible + cartaTapada

    print(f'El crupier mostra un {cartaVisible}')
    print(f'Tens {cartesUsuari}')

    while sortir == False:
        if cartesUsuari > 21:
            print("Has perdut! T'has passat de 21!")
            sortir = True

        elif cartesUsuari == 21:
            print('Has clavat 21!')
            sortir = True

        else:
            jugador = input('Demanar (P) o Plantar-se (S): ').upper()

            if jugador == 'P':
                cartesUsuari += random.randint(1, 10)
                print(f'Tens {cartesUsuari}')

            elif jugador == 'S':
                print(f'El crupier revela la carta tapada: {cartaTapada}')
                print(f'El crupier te {cartesCrupier}')

                while cartesCrupier < 17:
                    novaCarta = random.randint(1, 10)
                    cartesCrupier += novaCarta
                    print(f'El crupier roba un {novaCarta}')
                    print(f'El crupier te {cartesCrupier}')

                if cartesCrupier > 21:
                    print("Has guanyat! El crupier s'ha passat de 21!")
                elif cartesCrupier > cartesUsuari:
                    print('Has perdut!')
                elif cartesCrupier < cartesUsuari:
                    print('Has guanyat!')
                else:
                    print('Has empatat!')

                sortir = True

            else:
                print('Opcio no valida!')
```

```python
salir = False

while salir == False:
  print('''
        -- MENU --
  0 - Salir
  1 - Pedra Paper Tisores
  2 - El Penjat (Ahorcado)
  3 - Tres en ratlla
  4 - BlackJack''')
  try:
    opcion = int(input('Introdueix una opcio: '))

    match opcion:
      case 0:
        salir = True

      case 1:
        print('Has escollit pedra paper tisores!')
        pedraPaperTisora()

      case 2:
        print('Has escollit el penjat (Ahorcado)!')
        elPenjat()

      case 3:
        print('Has escollit el tres en ratlla!')
        tresEnRatlla()

      case 4:
        print('Has escollit el BlackJack!')
        blackJack()

  except ValueError:
    print('Has introduït un valor no valid')
  except KeyboardInterrupt:
    print('Finalitzat pel usuari')
    break
```
